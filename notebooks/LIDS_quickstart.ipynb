# LIDS Quick Start

See LIDS work immediately, then use `get_text()` to load and compare your own plain-text files. Scores range from `0` to `1`; larger values indicate stronger semantic similarity.

## 1. Setup

Install once from the repository root with `python -m pip install -r requirements.txt`, then run this notebook from the `notebooks` folder. The first run downloads `bert-base-uncased` and may take a moment.

```python
import os

# Move from the notebooks folder to the repository root.
os.chdir("..")

from lids import LIDS_similarity, get_text
```

## 2. LIDS in a nutshell

Put your texts in a list, choose the first as the reference, and compare every other text with it. Here, the two gardening summaries should score more highly than the unrelated spacecraft text.

```python
texts = [
    # 1. Reference text: longer reference
    (
        "Successful gardening requires regular attention to several important factors. "
        "Plants need healthy soil, sufficient sunlight, and consistent watering to grow properly throughout the season. "
        "Gardeners must also monitor their plants, remove weeds, and prune when necessary to prevent problems and maintain healthy growth. "
        "Adjusting care based on seasonal conditions can help keep a garden productive and healthy."
    ),

    # 2. Summary: similar terminology and phrasing
    "Successful gardening requires healthy soil, sufficient sunlight, consistent watering, and regular maintenance such as removing weeds and pruning.",

    # 3. Summary: different vocabulary, same information
    "A thriving garden depends on fertile ground, adequate light, reliable hydration, and routine upkeep to support plant growth throughout the year.",

    # 4. Completely unrelated
    "Researchers developed a new spacecraft navigation system that uses onboard sensors to calculate trajectories and make precise adjustments during long-distance missions.",
]

text_names = [
    "reference",
    "garden summary similar vocabulary",
    "garden summary different vocabulary",
    "unrelated to reference",
]

reference = texts[0]
for name, text in zip(text_names[1:], texts[1:]):
    score = LIDS_similarity(reference, text)
    print(f"reference vs {name}: {score:.4f}")
```

We see that both gardening summaries are more similar to the reference than the spacecraft passage. A major strength of LIDS is its ability to recognize shared meaning even when the vocabulary changes.

## 3. Load files with `get_text()`

`get_text(path)` reads a plain-text file and returns its complete contents as a Python string. Once the files are loaded, pass the two strings directly to `LIDS_similarity(text1, text2)`.

```python
# Note text1 and tex2 are the same  as the reference text and garden summary similar vocabulary text from above
text1 = get_text("examples/gardening_reference.txt")
text2 = get_text("examples/gardening_summary.txt")

score = LIDS_similarity(text1, text2)
print(f"LIDS similarity: {score:.4f}")
```

That is the complete file workflow. Replace the two paths with your own `.txt` files and rerun the cell.

```python
my_text1 = get_text("path/to/first_text.txt")
my_text2 = get_text("path/to/second_text.txt")
print(LIDS_similarity(my_text1, my_text2))
```
